## 1. Importer les outils de base

- `Path` sert à manipuler les chemins de fichiers.
- `subprocess` pour executer le script `eq.py` depuis le notebook.
- `pandas` sert à lire les fichiers CSV sous forme de tableau.

````python
from pathlib import Path
import subprocess
import sys

import pandas as pd
````

## 2. Trouver le dossier du projet

Le script `eq.py` cherche un fichier appelé `MainTable.csv` dans un dossier.

Par défaut, on part du principe qu'ils sont au même niveau.

````python
depart = Path.cwd()

# Le dossier ProgSnap2Analysis est au même niveau que le dossier JUPYTER.
PROJECT_DIR = (depart / "../ProgSnap2Analysis").resolve()

# Chemins importants du projet
MAIN_TABLE = PROJECT_DIR / "MainTable.csv"
EQ_SCRIPT = PROJECT_DIR / "eq.py"
OUTPUT_DIR = PROJECT_DIR / "out_eq_debutant"
OUTPUT_FILE = OUTPUT_DIR / "EQ.csv"

if not PROJECT_DIR.exists():
    raise FileNotFoundError(f"Le dossier du projet n'existe pas : {PROJECT_DIR}")

if not MAIN_TABLE.exists():
    raise FileNotFoundError(f"MainTable.csv est introuvable : {MAIN_TABLE}")

if not EQ_SCRIPT.exists():
    raise FileNotFoundError(f"eq.py est introuvable : {EQ_SCRIPT}")

print("\nOK : les fichiers nécessaires ont été trouvés.")
````

## 3. Lire `MainTable.csv`

Chargement du MainTable.csv

````python
main_table = pd.read_csv(MAIN_TABLE)

print("Nombre de lignes :", len(main_table))
print("Nombre de colonnes :", len(main_table.columns))

main_table.head()
````

## 4. Chercher les colonnes importantes pour `eq`

A modifier pour les scripts complets. Ou à diviser pour éxecuter chacun ?

````python
colonnes_eq = [
    "SubjectID",
    "Order",
    "EventType",
    "EventID",
    "ParentEventID",
    "CompileMessageType",
]

main_table[colonnes_eq].head(10)
````

## 5. Lancer `eq.py`

```bash
python eq.py . out_eq_debutant/EQ.csv
```
`eq.py` calcule l'**Error Quotient** pour chaque étudiant et écrit le résultat dans un fichier CSV.

````python
OUTPUT_DIR.mkdir(exist_ok=True)
LOG_FILE = OUTPUT_DIR / "execution_eq.log"

commande = [
    sys.executable,          # le Python utilisé par Jupyter
    str(EQ_SCRIPT),          # le script eq.py
    str(PROJECT_DIR),        # le dossier où se trouve MainTable.csv
    str(OUTPUT_FILE),        # le fichier CSV à créer
]

#Envoi du stdout vers un log pour plus de clarté.
with open(LOG_FILE, "w", encoding="utf-8") as log:
    resultat = subprocess.run(
        commande,
        cwd=PROJECT_DIR,
        text=True,
        stdout=log,
        stderr=subprocess.STDOUT,
    )

print("Code retour :", resultat.returncode)
print("Fichier log :", LOG_FILE)
print("Fichier résultat créé :", OUTPUT_FILE.exists())

# Check pour échec
resultat.check_returncode()
````

## 6. Lire le résultat `EQ.csv`
